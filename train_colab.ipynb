Este notebook treina a CNN customizada (`EsteiraCNN`) gratuitamente na nuvem (Google Colab), avalia no conjunto de teste (15% fotos reais) e faz o download do arquivo `cnn_esteira.pt` (~300 KB) pronto para hospedar a API FastAPI.

**Como usar:**
1. Escolha na célula abaixo se o projeto está no **GitHub** (recomendado) ou no **Google Drive**.

3. Ao final, o arquivo `cnn_esteira.pt` será baixado automaticamente — coloque-o em `model/cnn_esteira.pt` no repositório e faça o deploy!

```python
# === CONFIGURAÇÃO ===
# Escolha 'github' (recomendado) ou 'drive'
ORIGEM_DADOS = "github"  # @param ["github", "drive"]

# Se usar GitHub, coloque a URL do seu repositório:
GITHUB_REPO_URL = "https://github.com/SEU_USUARIO/SEU_REPOSITORIO.git"  # @param {type:"string"}

# Se usar Google Drive, coloque o caminho da pasta do projeto no seu Drive:
DRIVE_PROJECT_DIR = "/content/drive/MyDrive/projeto_harduino-ia"  # @param {type:"string"}

EPOCHS = 40  # @param {type:"integer"}
BATCH_SIZE = 8  # @param {type:"integer"}
LEARNING_RATE = 0.001  # @param {type:"number"}
```
